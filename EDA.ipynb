# NYC Airbnb Price Analysis
## Exploratory Data Analysis

This notebook performs initial data exploration on NYC Airbnb listings to understand pricing patterns, distributions, and key features.

```python
import wandb
import pandas as pd

run = wandb.init(project="nyc_airbnb", group="eda", save_code=True)
local_path = wandb.use_artifact("sample.csv:latest").file()
df = pd.read_csv(local_path)
df.head()
```

## Data Profiling

Generate a comprehensive statistical summary and automated data quality report using ydata profiling.

```python
import ydata_profiling

profile = ydata_profiling.ProfileReport(df)
profile.to_notebook_iframe()
```

## Data Cleaning & Preprocessing

- Filter price outliers: Remove $0 listings (data errors) and prices >$350/night (unrealistic)
- Convert `last_review` to datetime: Ydata profiling revealed high missing values; converting format helps identify and handle nulls properly
- Validate data structure and types

```python
# Drop outliers - Remove listings with unrealistic prices
# Filter out $0 listings (likely data errors) and extreme high prices (>$350/night)
min_price = 10
max_price = 350
idx = df['price'].between(min_price, max_price)
df=df[idx].copy()

# Convert last_review to datetime format for time-based analysis
df['last_review'] = pd.to_datetime(df['last_review'])
df.info()
run.finish()
```
